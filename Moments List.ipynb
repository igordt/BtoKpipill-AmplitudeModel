# Moments List - Esportazione dei Momenti Angolari

Questo notebook carica i momenti angolari calcolati (salvati come file .pkl nella cartella WignerAmplitude)
e li esporta in file di testo separati all'interno della cartella 'Moments_Output', divisi per simmetria (ω = +1 o ω = -1).
Ogni momento viene formattato in modo leggibile e salvato in un file singolo.

```python
'''
Caricamento dei dati.
Importa le librerie necessarie e carica i dati pre-calcolati dei momenti dalla cartella WignerAmplitude.
'''

import winsound
import sympy as sp
import dill
import os

sp.init_printing(use_unicode=True, wrap_line=True, num_columns=150)

# Percorso della cartella WignerAmplitude
wigner_dir = os.path.join(os.getcwd(), "WignerAmplitude")

# Percorsi completi dei file
sorted_keys_file = os.path.join(wigner_dir, "sorted_keys.pkl")
moments_file = os.path.join(wigner_dir, "Moments.pkl")
symmetry_file = os.path.join(wigner_dir, "symmetry.pkl")

# Carica sorted_keys
print("Caricamento sorted_keys.pkl...")
with open(sorted_keys_file, 'rb') as f:
    sorted_keys = dill.load(f)
    print(f"sorted_keys caricato: {len(sorted_keys)} elementi")

# Carica Moments
print("\nCaricamento Moments.pkl...")
with open(moments_file, 'rb') as f:
    moments = dill.load(f)
    print(f"Moments caricato")

# Carica Symmetry
print("\nCaricamento Symmetry.pkl...")
with open(symmetry_file, 'rb') as f:
    symmetry = dill.load(f)
    print(f"symmetry caricato")
```

```python
'''
Esportazione dei momenti in file di testo.
Salva ogni momento in un file di testo separato, organizzato per simmetria.
'''

def pretty_terms(expr, num_columns):
    """
    Restituisce una stringa con un termine per riga.
    """
    lines = []
    
    if isinstance(expr, sp.Add):
        for i, term in enumerate(expr.as_ordered_terms(), 1):
            lines.append(
                sp.pretty(term, wrap_line=True, num_columns=num_columns)
            )
    else:
        lines.append(
            sp.pretty(expr, wrap_line=True, num_columns=num_columns)
        )
    
    return "\n".join(lines)

# Crea la cartella per salvare gli output
output_dir = os.path.join(os.getcwd(), "Moments_Output")
os.makedirs(output_dir, exist_ok=True)
print(f"Cartella di output: {output_dir}\n")

count = 0
for k, key in enumerate(sorted_keys):
    if symmetry[key] == 1:
        count += 1
        
        # Prepara il contenuto da salvare
        output_content = []
        output_content.append("="*50)
        output_content.append(f"D_{count:03d}, ω = 1:")
        output_content.append("\nKey:")
        output_content.append(sp.pretty(key))
        output_content.append("\nMoment:")
        output_content.append(
            pretty_terms(moments[key], num_columns=200)
        )
        output_content.append("="*50)
        
        # Salva in un file
        filename = f"D_{count:03d}_omega_piu.txt"
        filepath = os.path.join(output_dir, filename)
        
        with open(filepath, 'w', encoding='utf-8') as f:
            f.write('\n'.join(output_content))

for k, key in enumerate(sorted_keys):
    if symmetry[key] == -1:
        count += 1
        
        # Prepara il contenuto da salvare
        output_content = []
        output_content.append("="*50)
        output_content.append(f"D_{count:03d}, ω = -1:")
        output_content.append("\nKey:")
        output_content.append(sp.pretty(key))
        output_content.append("\nMoment:")
        output_content.append(
            pretty_terms(moments[key], num_columns=200)
        )
        output_content.append("="*50)
        
        # Salva in un file
        filename = f"D_{count:03d}_omega_meno.txt"
        filepath = os.path.join(output_dir, filename)
        
        with open(filepath, 'w', encoding='utf-8') as f:
            f.write('\n'.join(output_content))
        
        
print(f"\nCompletato! {count} file salvati in '{output_dir}'")
winsound.Beep(1000, 500)
```
